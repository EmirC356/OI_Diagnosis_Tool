```python
import pandas as pd;
import matplotlib.pyplot as plt;
import Bio as bio;
```

```python

All ={}
for record in bio.parse("file.fasta", "fasta"):
    All[record.id]= record.seq

print(All)
```

```python
def SeqStats(id):
        if id not in All:
                print("Id is not in the file. \n")
                return
        Name = id
        Seq = All[id]
        Length = len(Seq)
        Ncount = Seq.count('N')
        Gcount = Seq.count('G')
        Ccount = Seq.count('C')
        Acount = Seq.count('A')
        Tcount = Seq.count('T')

        GCperc = (Gcount+Ccount)*100/ Length
        ATskew = (Acount-Tcount) / (Acount+Tcount)
        GCskew = (Gcount-Ccount) / (Gcount+Ccount)

        return{
        "name": Name,
        "length": Length,
        "n_count": Ncount,
        "gc_percent": round(GCperc, 2),
        "at_skew": round(ATskew, 3),
        "gc_skew": round(GCskew, 3),
        }


stats = SeqStats("name2")
print(stats)
```

```python
def Transcript(id):
    if id not in All:
                print("Id is not in the file. \n")
                return
    
    Seq = All[id].upper()
    return Seq.replace("T", "U")
    
print(Transcript("name2"))
```

```python
def RevComp(id):
    if id not in All:
                print("Id is not in the file. \n")
                return
    Seq = All[id]
    Comp = ""
    for i in range(len(Seq)):

        if Seq[i] == 'T': Comp += ('A') 
        elif Seq[i] == 'A': Comp += ('T') 
        elif Seq[i] == 'G': Comp += ('C') 
        elif Seq[i] == 'C': Comp += ('G') 

        else: Comp += Seq[i]
    return Comp[::-1]

print(RevComp("name1"))
```

```python
def find_motif(id, motif):
     if id not in All:
                print("Id is not in the file. \n")
                return
     motif=motif.upper()
     seq = All[id].upper()
     j = len(seq)
     l = len(motif)
     indexs = []
     for i in range(j-l+1):
             if seq[i:i+l] == motif: indexs.append(i)
     return(indexs)
print(find_motif("name3", "AGC"))
```

```python
def find_motif_miss(id,motif,miss, use_iupac:bool = True):
    if id not in All:
                print("Id is not in the file. \n")
                return
    motif=motif.upper()
    seq = All[id].upper()
    j = len(seq)
    l = len(motif)
    indexs = []
    for i in range(j-l+1):
             count =0
             for x in range(l):
                if seq[i:i+l][x] != motif[x]: count +=1 
                if count > miss: break
             if count <= miss: indexs.append(i)
    return indexs

print(find_motif_miss("name3", "GAC",1))
```

```python
def k_mer(id,k):
    if id not in All:
                print("Id is not in the file. \n")
                return
    seq = str(All[id]).upper()
    result ={}
    for i in range(len(seq)-k+1):
        kmer = seq[i:i+k]
        if set(kmer) <= {"A","C","G","T"}:

          if kmer in result: result[kmer] += 1
          else: result[kmer] = 1
        else:
            continue
    sorted_result = sorted(result.items(), key=lambda x: x[1], reverse=True)    
    sorted_result =  dict(sorted_result)

    return sorted_result
print(k_mer("name2",3))
```
