```python
import pandas as pd
import numpy as np
from Bio import SeqIO

```

```python

codon_table = {
    'ATA':'I', 'ATC':'I', 'ATT':'I', 'ATG':'M',
    'ACA':'T', 'ACC':'T', 'ACG':'T', 'ACT':'T',
    'AAC':'N', 'AAT':'N', 'AAA':'K', 'AAG':'K',
    'AGC':'S', 'AGT':'S', 'AGA':'R', 'AGG':'R',
    'CTA':'L', 'CTC':'L', 'CTG':'L', 'CTT':'L',
    'CCA':'P', 'CCC':'P', 'CCG':'P', 'CCT':'P',
    'CAC':'H', 'CAT':'H', 'CAA':'Q', 'CAG':'Q',
    'CGA':'R', 'CGC':'R', 'CGG':'R', 'CGT':'R',
    'GTA':'V', 'GTC':'V', 'GTG':'V', 'GTT':'V',
    'GCA':'A', 'GCC':'A', 'GCG':'A', 'GCT':'A',
    'GAC':'D', 'GAT':'D', 'GAA':'E', 'GAG':'E',
    'GGA':'G', 'GGC':'G', 'GGG':'G', 'GGT':'G',
    'TCA':'S', 'TCC':'S', 'TCG':'S', 'TCT':'S',
    'TTC':'F', 'TTT':'F', 'TTA':'L', 'TTG':'L',
    'TAC':'Y', 'TAT':'Y', 'TAA':'_', 'TAG':'_',
    'TGC':'C', 'TGT':'C', 'TGA':'_', 'TGG':'W',
}
def translate_dna_to_protein(dna_sequence):
    protein_sequence = ""
    for i in range(0, len(dna_sequence), 3):
        codon = dna_sequence[i:i+3]
        amino_acid = codon_table.get(codon, 'X') 
        if amino_acid == '_':
            return protein_sequence 
        protein_sequence += amino_acid
    return protein_sequence     

seq = "ATGGCCATTGTAATGGGCCGCTGAAAGGGTGACCGATAG"
protein = translate_dna_to_protein(seq)
print("Protein Sequence:", protein)

```

```python
def all_snps(seq):
    bases = ['A', 'T', 'C', 'G']
    snp_list=[]
    for i in range(len(seq)):
        for b in bases:
            if seq[i] != b:
                snp = seq[:i] + b +  seq[i+1:]
                snp_list.append(snp)
    return snp_list
snp_variants = all_snps(seq)
print("Number of SNP variants generated:", len(snp_variants))
seq = "ATGCTAGAT"
seq_vars = all_snps(seq)
print("SNP Variants for sequence", seq, ":", seq_vars)
print("Total SNP Variants:", len(seq_vars))
```

```python
def all_genes(file_name):
  
    All ={}
    for record in SeqIO.parse(file_name, "fasta"):
     All[record.id]= record.seq

    return All
```

```python
def test_harmfullnessCOL1A1(orj_prot,mut_prot):
    #HARMFULNESS CRITERIA if GLY breaks or NONSENSE mutations

    for i in range(min(len(orj_prot),len(mut_prot))):
        if orj_prot[i] == 'G' and mut_prot[i] != 'G':
            return True
        if mut_prot[i] == '_' and orj_prot[i] != '_':
            return True

    return False
def harmful_snps(seq):
    snp_list = all_snps(seq)
    harmful_list = []
    original_protein = translate_dna_to_protein(seq)
    for snp in snp_list:
        mutated_protein = translate_dna_to_protein(snp)
        if test_harmfullnessCOL1A1(original_protein, mutated_protein):
            harmful_list.append(snp)
    
    return harmful_list
All = all_genes("gene.fna")
Sequence_Names = list(All.keys())
print(Sequence_Names)
for name in Sequence_Names:

    Sequence = All[name]
    print(name)
   
    print(Sequence[:20])
    harmful_variants = harmful_snps(str(Sequence))
    print("Number of harmful SNP variants generated:", len(harmful_variants))

```
